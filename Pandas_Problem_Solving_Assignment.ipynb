## **Load and Inspect the Data**

```python
import pandas as pd
Data = {
    "EmpID":[101,102,103,104,105,106,107,108,109,110],
    "Name":['Arun','Priya','Kumar','Divya','Rahul','Meena','Sathish','Nila','Vijay','Kavya'],
    "Department":['IT','HR','IT','Finance','IT','HR','Finance','IT','Sales','Sales'],
    "Age":[25,29,32,27,24,35,31,28,30,26],
    "Salary":[35000,42000,58000,39000,30000,62000,51000,47000,44000,36000],
    "Experience":[2,5,8,3,1,10,7,4,6,2],
    "City":['Chennai','Madurai','Chennai','Coimbatore','Salem','Chennai','Madurai','Coimbatore','Chennai','Salem']
}

df = pd.DataFrame(Data)

print("First 5 rows:")
print(df.head())

print("\nLast 3 rows:")
print(df.tail(3))

print("\nNumbers of rows and columns:")
print(df.shape)

print("\nColumn Data Types:")
print(df.dtypes)

```

## **Filter Employee Records**

```python
Filtered_df = df[(df["Department"]=="IT")&(df["Salary"]> 4000)]

print(Filtered_df)
```

## **Sort the Data**

```python
Sorted_df = df.sort_values(by="Salary",ascending = False)

print(Sorted_df[["Name","Department","Salary"]])
```

## **Calculate Salary Statistics**

```python
print("Mean Salary:",df["Salary"].mean())
print("Maximum Salary:",df["Salary"].max())
print("Minimum Salary:",df["Salary"].min())
print("Median Salary:",df["Salary"].median())

print("Average Age:",df["Age"].mean())
```

## **Group By Department**

```python
result = df.groupby("Department").agg(Number_of_Employees = ("EmpID","count"),
                                      Average_Salary = ("Salary","mean"),
                                      Highest_Salary = ("Salary","max"))

order = result.sort_values(by="Average_Salary",ascending=False)

print(order)
                        
        
```

## **Create a New Column Annual Salary**

```python
df["Annual Salary"] = df["Salary"]*12

print(df[["Name","Salary","Annual Salary"]])
```

## **Find Misssing and Duplicate Values**

```python
new_df = df.copy()

new_df.loc[0,"City"] = None

new_df = pd.concat([new_df,new_df.iloc[[1]]],ignore_index = True)

print("Missing Values:")
print(new_df.isnull().sum())

new_df["City"] = new_df["City"].fillna("Unknown")

print("\nNumber of Duplicate row:")
print(new_df.duplicated().sum())

nwe_df = new_df.drop_duplicates()

print("\nCleaned DataFrame:")
print(new_df)
```

## **Use Conditional Logic**

```python
import numpy as np

df["Salary Level"] = np.where(df["Salary"]>=50000,"High",np.where(df["Salary"]>=40000,"Median","Low"))

print(df[["Name","Salary","Salary Level"]])
```

## **Work with a second DataFrame Merge**

```python
import pandas as pd

marks = pd.DataFrame({"EmpID":[101,103,105,108,110],
                      "Trainning score":[85,90,78,88,92]})
merged_df = pd.merge(df,marks,on="EmpID",how="left")

print(merged_df,[["EmpID","Name","Trainning score"]])
```

## **Answer a Business Questions**

```python
avg_Salary = df.groupby("Department")["Salary"].mean()

highest_department = avg_Salary.idxmax()
highest_avg_Salary = avg_Salary.max()

print("Department with Highest Average Salary:",highest_department)
print("Highest Average Salary:",highest_avg_Salary)

overall_avg_Salary = df["Salary"].mean()

print("\nOverall Average Salary:",overall_avg_Salary)

above_avg = df[df["Salary"]>overall_avg_Salary]

print("\nEmployees Earning Above Average Salary:")
print(above_avg[["Name","Department","Salary"]])
```
