```python
x = [1,2,3,4,5]
y1 = [10,20,30,40,45]
y2 = [15,20,40,35,50]

plt.plot(x,y1,label="Students",marker = "*")
plt.plot(x,y2,label="Teachers",marker = "^")

plt.title("Students vs Teachers")
plt.xlabel("X Values")
plt.ylabel("Y Values")

plt.legend()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
fig,ax = plt.subplots(figsize = (10,6))

scatter = ax.scatter(df["age"],df["trestbps"],c = df["target"],cmap = "winter")
ax.set(title = "Age and TrestBPS Levels",xlabel = "Age",ylabel = "TrestBPS")
ax.legend(*scatter.legend_elements(),title = "Target")
ax.axhline(df["trestbps"].mean(),c = "r",linestyle = "dashed");


plt.show();
          
```

```python
fig,ax = plt.subplots(figsize = (15,10))

ax.bar(df["age"],df["trestbps"],color = "skyblue",edgecolor = "black")

ax.set(title = "Age and Trestbps Levels",xlabel = "Age",ylabel = "TrestBPS")

ax.axhline(df["trestbps"].mean(),color= "r",linestyle = "dashed",label = "Mean TrestBPS")

ax.legend()

plt.show();
```

```python
fig,ax = plt.subplots(figsize = (15,6))

ax.hist(df["age"],bins = 20,color = "skyblue",edgecolor = "black",alpha = 0.8)
ax.hist(df["trestbps"],bins = 35,color = "orange",edgecolor = "black",alpha = 1)

ax.axhline(df["age"].mean(),color = "green",linestyle = "dashed",label = "Age")
ax.axhline(df["trestbps"].mean(),color= "r",linestyle = "dashed",label = "Mean TrestBPS")

ax.set(title = "Age and Trestbps Levels",xlabel = "Age",ylabel = "TrestBPS")

ax.legend()
plt.show();
```
